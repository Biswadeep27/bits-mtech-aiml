```python
import os
from openai import OpenAI 
from IPython.display import display, HTML
```

```python
dbutils.widgets.text("workspace_url", "")
dbutils.widgets.text("base_llm_model", "agents_beepz_dissertation_draft-bronze_draft-querymorph_sqlgen")
ws_url = dbutils.widgets.get("workspace_url")
base_url = f"{ws_url.rstrip('/')}//serving-endpoints"
base_llm_model = dbutils.widgets.get("base_llm_model")
embedding_model = "databricks-gte-large-en"
```

```python
#Env Setup for OpenAI token
DATABRICKS_TOKEN = dbutils.notebook.entry_point.getDbutils().notebook().getContext().apiToken().get()

client = OpenAI(
  api_key=DATABRICKS_TOKEN,
  base_url=base_url
)
```

```python
def get_completion(prompt: str, model: str = base_llm_model):
  completions = client.chat.completions.create(
    model = model,
    #store = True,
    temperature = 0,
    messages = [
      {"role": "user", "content": prompt}
    ],
    
  )
  return completions.choices[0].message.content
```

```python
prompt = f"""
Give me 5 queries on Snowflake.
"""
response = get_completion(prompt)
print(response)
```

```python
prompt = f"""
give me 10 Snowflake SQL queries that has left joins, use of date functions and aggregates. Provide the output in an array format of string SQL values.
"""
response = get_completion(prompt)
print(response)
```
